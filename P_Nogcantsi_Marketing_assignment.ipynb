## Merge the ecommerce and US holidays datasets

```python
#Convert the string date columns in both datasets to date format
```

```python
ecommerce_df["new_date"] = pd.to_datetime(ecommerce_df["Date"])
ecommerce_df = ecommerce_df.drop(["Date"], axis=1)
```

```python
us_holiday_dates["new_date"] = pd.to_datetime(us_holiday_dates["Date"])
us_holiday_dates = us_holiday_dates.drop(["Date"], axis=1)
```

```python
merged_df = pd.merge(ecommerce_df, us_holiday_dates, how="inner", on="new_date")
```

```python
merged_df.shape
```

```python
merged_df.head(2)
```

```python
merged_df.rename(columns={"new_date": "Date"}, inplace = True)
```

## Create relevant new variables from existing ones

```python
merged_df["revenue"] = merged_df["Quantity"] * merged_df["UnitPrice"]
```

```python
merged_df.head(2)
```

```python
col_reorg = ['Country','StockCode', 'CustomerID', 'InvoiceNo',
             'Description', 'Date', 'Holiday','WeekDay', 'Hour',
             'Month', 'Day', 'Year','Quantity', 'UnitPrice', 'revenue']
```

```python
merged_df = merged_df[col_reorg]
```

```python
merged_df.head(2)
```

## Exploratory Analysis

```python
# Number and proportion of sales by country
cnt = merged_df["Country"].value_counts().to_frame()
cnt.rename(columns = {"Country":"count"}, inplace = True)
prop =round(merged_df["Country"].value_counts(normalize=True)*100, 2).to_frame()
prop.rename(columns = {"Country":"percent"}, inplace = True)
pd.merge(cnt, prop, left_index=True, right_index=True)
```

```python
# Top 10 item description
pd.DataFrame(merged_df['Description'].value_counts()[:10])
```

```python

# By default the discrete values along axis are ordered alphabetically.
# If we want a specific ordering we use a pandas.Categorical variable with categories ordered to our preference.

# Determine order and create a categorical type
# Note that value_counts() is already sorted
holiday_list = merged_df["Holiday"].value_counts().index.tolist()[::-1]
holiday_cat = pd.Categorical(merged_df["Holiday"], categories=holiday_list)

# assign to a new column in the DataFrame
merged_df = merged_df.assign(holiday_cat = holiday_cat)

(ggplot(merged_df)
 + aes(x='holiday_cat', fill= "Holiday")
 + geom_bar(size=20)
 + coord_flip()
 + labs(y='Count', x='Holidays', title='Thanksgiving Eve had the highest count')
 + theme_bw()
)
```

```python
# Frequency by day of week
sales_by_weekday = pd.DataFrame(merged_df["WeekDay"].value_counts())
sales_by_weekday.rename(columns = {"WeekDay":"Number_of_Sales_by_WeekDay"})
```

```python
merged_df = merged_df.loc[merged_df["Quantity"]>0, ]
```

```python

# Summary for revenue
pd.DataFrame(merged_df[["Quantity", "UnitPrice", "revenue"]].describe())
```

```python
#Quantity sold by holiday
rev_by_holiday = merged_df.groupby("Holiday")["Quantity"].agg(["count","sum", "mean", "min", "max", "std"])
rev_by_holiday.sort_values("sum", ascending = False)
```

```python
#Revenue by holiday
rev_by_holiday = merged_df.groupby("Holiday")["revenue"].agg(["count","sum", "mean", "min", "max", "std"])
rev_by_holiday.sort_values("sum", ascending = False)
```

```python
(
    ggplot(merged_df)
    + geom_boxplot(aes(x='factor(Holiday)', y='revenue'))
    + ylim(5, 50)
    + coord_flip()
)
```

```python
(
    ggplot(merged_df)
    + geom_violin(aes(x='factor(Holiday)', y='revenue'))
    + ylim(5, 50)
    + coord_flip()
)
```

```python
#Revenue by holiday
rev_by_holiday = merged_df.groupby(["Holiday"])["revenue"].agg(["count","sum", "mean", "min", "max", "std"])
rev_by_holiday["prop"] = round((rev_by_holiday['sum']/rev_by_holiday['sum'].sum()),4)
rev_by_holiday.sort_values(["sum"], ascending = False)
# According to the table below, sales on Thanksgiving Eve surpass those on other holidays.
# On this particular day, a total of 665 sales were made, generating $13,587.13 in revenue.
# Notably, this represents approximately 18.8% of the total revenue generated thus far.
```

```python
rev_by_country = merged_df.groupby(["Country"])["revenue"].agg(["count","sum", "mean", "min", "max", "std"])
rev_by_country["prop"] = round((rev_by_country['sum']/rev_by_country['sum'].sum()),4)
rev_by_country.sort_values(["sum"], ascending = False)
# Approximately 77% of the total revenue is generated from the United Kingdom.
```
